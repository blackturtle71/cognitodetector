```python
import pandas as pd
from collections import Counter
import numpy as np

p = "/home/gabs/Documents/lab-NN/cognitive_impairments/data/DementiaBank/parsed_data_bge-base-en-v1.5.pkl"
```

```python
df = pd.read_pickle(p)

# choose cookie only
df = df[df['ID'].str.contains(r'cookie$', regex=True)]
# leave patient id only
df['ID'] = df['ID'].str.split('-').str[0]
# drop duplicate IDs, we care about their age and sex only
df = df.drop_duplicates(subset=['ID'])
```

```python
df
```

```python
sex_count_con = Counter(df[df['dementia'] == 'Control']['sex'])
sex_count_dem = Counter(df[df['dementia'] != 'Control']['sex'])

print(f"{'#'*5}AGE-DISTRO{'#'*5}")
print(f"Min Control Age: {min(df[df['dementia'] == 'Control']['age'])}")
print(f"Mean Control Age: {round(np.mean(df[df['dementia'] == 'Control']['age']))}")
print(f"Max Control Age: {max(df[df['dementia'] == 'Control']['age'])}")
print("#"*20)
print(f"Min Dementia Age: {min(df[df['dementia'] != 'Control']['age'])}")
print(f"Mean Dementia Age: {round(np.mean(df[df['dementia'] != 'Control']['age']))}")
print(f"Max Dementia Age: {max(df[df['dementia'] != 'Control']['age'])}")
print("#"*20)
print()
print(f"{'#'*5}SEX-DISTRO{'#'*5}")
print(f"Num Females in Control: {sex_count_con['female']}")
print(f"Num Males in Control: {sex_count_con['male']}")
print("#"*20)
print(f"Num Females in Dementia: {sex_count_dem['female']}")
print(f"Num Males in Dementia: {sex_count_dem['male']}")
print("#"*20)
```
