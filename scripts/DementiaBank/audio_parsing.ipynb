```python
import whisper
from whisper.utils import get_writer
from pydub import AudioSegment
from typing import List, Dict
```

```python
class SpeechToText:
    def __init__(self, model_size='base'):
        self.model = whisper.load_model(model_size)
        
    def transcribe(self, audio_path):
        result = self.model.transcribe(audio_path)
        return result['text']
    
transcriber = SpeechToText()
```

```python
transcript = transcriber.transcribe('/home/gabs/Documents/lab-NN/cognitive_impairments/data/manual_recordings/audio/test_control/test_control_audio.wav')
with open('/home/gabs/Documents/lab-NN/cognitive_impairments/data/manual_recordings/transcript/control_transcript_test.txt', "w") as fout:
    fout.write(transcript)
```
